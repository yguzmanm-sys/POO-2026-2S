```python
from math import sqrt
import ipywidgets as widgets
from IPython.display import display

class Notas:
    def __init__(self, tamanio=5):
        self.lista_notas = [0.0] * tamanio

    def calcular_promedio(self):
        return sum(self.lista_notas) / len(self.lista_notas) if self.lista_notas else 0.0

    def calcular_desviacion(self):
        if not self.lista_notas:
            return 0.0
        promedio = self.calcular_promedio()
        suma_cuadrados = sum((n - promedio) ** 2 for n in self.lista_notas)
        return sqrt(suma_cuadrados / len(self.lista_notas))

    def calcular_menor(self):
        return min(self.lista_notas) if self.lista_notas else 0.0

    def calcular_mayor(self):
        return max(self.lista_notas) if self.lista_notas else 0.0

class InterfazColab:
    def __init__(self):
        self.notas = Notas()
        self.campos_notas = []
        self._crear_componentes()

    def _crear_componentes(self):
        # Crear campos de entrada para las 5 notas
        self.campos_notas = [
            widgets.FloatText(value=0.0, description=f'Nota {i + 1}:', style={'description_width': 'initial'})
            for i in range(5)
        ]

        # Botones
        self.btn_calcular = widgets.Button(description="Calcular", button_style='primary')
        self.btn_limpiar = widgets.Button(description="Limpiar", button_style='warning')

        # Etiquetas de salida
        self.lbl_promedio = widgets.Label(value="Promedio = ")
        self.lbl_desviacion = widgets.Label(value="Desviacion = ")
        self.lbl_mayor = widgets.Label(value="Nota mayor = ")
        self.lbl_menor = widgets.Label(value="Nota menor = ")
        self.lbl_error = widgets.Label(value="", style={'font_color': 'red'})

        # Eventos
        self.btn_calcular.on_click(self._calcular_resultados)
        self.btn_limpiar.on_click(self._limpiar_campos)

        # Contenedores de diseño (Layout)
        inputs_box = widgets.VBox(self.campos_notas)
        buttons_box = widgets.HBox([self.btn_calcular, self.btn_limpiar])
        outputs_box = widgets.VBox([
            self.lbl_promedio,
            self.lbl_desviacion,
            self.lbl_mayor,
            self.lbl_menor,
            self.lbl_error
        ])

        # Mostrar la interfaz completa
        display(widgets.VBox([
            widgets.HTML("<h3>Gestor de Notas</h3>"),
            inputs_box,
            buttons_box,
            outputs_box
        ]))

    def _obtener_notas(self):
        try:
            for i, campo in enumerate(self.campos_notas):
                valor = float(campo.value)
                self.notas.lista_notas[i] = valor
            self.lbl_error.value = ""
            return True
        except Exception:
            self.lbl_error.value = "Error: Todos los campos deben contener números válidos."
            return False

    def _calcular_resultados(self, _):
        if not self._obtener_notas():
            return
        self.lbl_promedio.value = f"Promedio = {self.notas.calcular_promedio():.2f}"
        self.lbl_desviacion.value = f"Desviacion = {self.notas.calcular_desviacion():.2f}"
        self.lbl_mayor.value = f"Nota mayor = {self.notas.calcular_mayor():.2f}"
        self.lbl_menor.value = f"Nota menor = {self.notas.calcular_menor():.2f}"

    def _limpiar_campos(self, _):
        for campo in self.campos_notas:
            campo.value = 0.0
        self.lbl_promedio.value = "Promedio = "
        self.lbl_desviacion.value = "Desviacion = "
        self.lbl_mayor.value = "Nota mayor = "
        self.lbl_menor.value = "Nota menor = "
        self.lbl_error.value = ""

# Ejecutar la interfaz interactiva en Google Colab
app = InterfazColab()
```

```python
import math
import ipywidgets as widgets
from IPython.display import display

class FiguraGeometrica:
    """Clase base que almacena el volumen y la superficie de una figura."""
    def __init__(self) -> None:
        self.volumen: float = 0.0
        self.superficie: float = 0.0

    def set_volumen(self, volumen: float) -> None:
        self.volumen = volumen

    def set_superficie(self, superficie: float) -> None:
        self.superficie = superficie

    def get_volumen(self) -> float:
        return self.volumen

    def get_superficie(self) -> float:
        return self.superficie


class Cilindro(FiguraGeometrica):
    """Representa un cilindro definido por su radio y altura."""
    def __init__(self, radio: float, altura: float) -> None:
        super().__init__()
        self.radio = radio
        self.altura = altura
        self.set_volumen(self.calcular_volumen())
        self.set_superficie(self.calcular_superficie())

    def calcular_volumen(self) -> float:
        return math.pi * self.altura * self.radio ** 2

    def calcular_superficie(self) -> float:
        area_lateral = 2 * math.pi * self.radio * self.altura
        area_bases = 2 * math.pi * self.radio ** 2
        return area_lateral + area_bases


class Esfera(FiguraGeometrica):
    """Representa una esfera definida unicamente por su radio."""
    def __init__(self, radio: float) -> None:
        super().__init__()
        self.radio = radio
        self.set_volumen(self.calcular_volumen())
        self.set_superficie(self.calcular_superficie())

    def calcular_volumen(self) -> float:
        return (4.0 / 3.0) * math.pi * self.radio ** 3

    def calcular_superficie(self) -> float:
        return 4.0 * math.pi * self.radio ** 2


class Piramide(FiguraGeometrica):
    """Representa una piramide de base cuadrada con apotema."""
    def __init__(self, base: float, altura: float, apotema: float) -> None:
        super().__init__()
        self.base = base
        self.altura = altura
        self.apotema = apotema
        self.set_volumen(self.calcular_volumen())
        self.set_superficie(self.calcular_superficie())

    def calcular_volumen(self) -> float:
        return (self.base ** 2) * self.altura / 3.0

    def calcular_superficie(self) -> float:
        area_base = self.base ** 2
        area_lados = 2.0 * self.base * self.apotema
        return area_base + area_lados


class FiguraAppColab:
    """Interfaz gráfica interactiva para Google Colab."""
    def __init__(self) -> None:
        self._crear_interfaz()

    def _crear_interfaz(self) -> None:
        self.selector_figura = widgets.Dropdown(
            options=['Cilindro', 'Esfera', 'Piramide'],
            value='Cilindro',
            description='Figura:',
        )

        # Campos de entrada individuales
        self.txt_radio = widgets.FloatText(value=0.0, description='Radio (cm):')
        self.txt_altura = widgets.FloatText(value=0.0, description='Altura (cm):')
        self.txt_base = widgets.FloatText(value=0.0, description='Base (cm):')
        self.txt_apotema = widgets.FloatText(value=0.0, description='Apotema (cm):')

        # Botones
        self.btn_calcular = widgets.Button(description="Calcular", button_style='primary')
        self.btn_limpiar = widgets.Button(description="Limpiar", button_style='warning')

        # Salidas
        self.lbl_volumen = widgets.Label(value="Volumen =")
        self.lbl_superficie = widgets.Label(value="Superficie =")
        self.lbl_error = widgets.Label(value="", style={'font_color': 'red'})

        # Contenedor dinámico de campos de texto
        self.campos_container = widgets.VBox([])

        # Eventos
        self.selector_figura.observe(self._actualizar_campos, names='value')
        self.btn_calcular.on_click(self.calcular)
        self.btn_limpiar.on_click(self.limpiar)

        # Inicializar y mostrar
        self._actualizar_campos(None)
        display(widgets.VBox([
            widgets.HTML("<h3>Cálculo de Figuras Geométricas</h3>"),
            self.selector_figura,
            self.campos_container,
            widgets.HBox([self.btn_calcular, self.btn_limpiar]),
            widgets.VBox([self.lbl_volumen, self.lbl_superficie, self.lbl_error])
        ]))

    def _actualizar_campos(self, change) -> None:
        figura = self.selector_figura.value
        if figura == "Cilindro":
            self.campos_container.children = [self.txt_radio, self.txt_altura]
        elif figura == "Esfera":
            self.campos_container.children = [self.txt_radio]
        else:  # Piramide
            self.campos_container.children = [self.txt_base, self.txt_altura, self.txt_apotema]

    def calcular(self, _):
        figura = self.selector_figura.value
        try:
            self.lbl_error.value = ""
            if figura == "Cilindro":
                r = float(self.txt_radio.value)
                h = float(self.txt_altura.value)
                if r <= 0 or h <= 0:
                    raise ValueError("Los valores deben ser mayores a cero.")
                figura_obj = Cilindro(r, h)
            elif figura == "Esfera":
                r = float(self.txt_radio.value)
                if r <= 0:
                    raise ValueError("El radio debe ser mayor a cero.")
                figura_obj = Esfera(r)
            else:
                b = float(self.txt_base.value)
                h = float(self.txt_altura.value)
                a = float(self.txt_apotema.value)
                if b <= 0 or h <= 0 or a <= 0:
                    raise ValueError("Los valores deben ser mayores a cero.")
                figura_obj = Piramide(b, h, a)

            self.lbl_volumen.value = f"Volumen = {figura_obj.get_volumen():.2f} cm^3"
            self.lbl_superficie.value = f"Superficie = {figura_obj.get_superficie():.2f} cm^2"
        except ValueError as e:
            self.lbl_error.value = f"Error: {str(e)}"

    def limpiar(self, _):
        self.txt_radio.value = 0.0
        self.txt_altura.value = 0.0
        self.txt_base.value = 0.0
        self.txt_apotema.value = 0.0
        self.lbl_volumen.value = "Volumen ="
        self.lbl_superficie.value = "Superficie ="
        self.lbl_error.value = ""

if __name__ == "__main__":
    app = FiguraAppColab()
```
